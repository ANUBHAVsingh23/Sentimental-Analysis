## Data preparation

### Subtask:
Prepare the data for sentiment analysis.  Handle missing values in the text column intended for sentiment analysis (if any exist), and convert the text to lowercase. Ensure the 'age' column is numeric and the 'gender' column is consistent.


**Reasoning**:
Prepare the data for sentiment analysis by handling missing values, converting text to lowercase, processing the age column, and standardizing the gender column.



```python
import pandas as pd
import numpy as np

# 1. Identify the text column (assuming it's 'Have you tried to get rid of your social media addiction? ')
text_column = 'Have you tried to get rid of your social media addiction? '

# 2. Handle missing values in the text column
if df[text_column].isnull().any():
    df[text_column] = df[text_column].fillna("")

# 3. Convert text to lowercase
df[text_column] = df[text_column].str.lower()


# 4. Convert the 'What is your age?' column to numeric
def convert_age(age_range):
    try:
      if pd.isna(age_range):
        return np.nan
      if 'to' in str(age_range):
          start, end = map(int, age_range.split(' to '))
          return (start + end) // 2  # Return the midpoint
      else:
        return int(age_range)
    except:
        return np.nan

df['What is your age?'] = df['What is your age?'].apply(convert_age)
df.dropna(subset=['What is your age?'], inplace=True)
df['What is your age?'] = df['What is your age?'].astype(int)


# 5. Standardize gender categories
gender_mapping = {
    'Male': 'Male',
    'Female': 'Female',
    'male': 'Male',
    'female': 'Female',
}
df['What is your Gender'] = df['What is your Gender'].replace(gender_mapping)
df['What is your Gender'].fillna("Other", inplace=True)


# 6. Check for missing values again
print(df[['What is your age?', 'What is your Gender', text_column]].isnull().sum())

display(df.head())
```

## Data analysis

### Subtask:
Perform sentiment analysis and group the results by age and gender.


**Reasoning**:
Perform sentiment analysis using TextBlob, calculate the sentiment score, group the data by age and gender, and calculate the average sentiment score for each group.



```python
from textblob import TextBlob

# Perform sentiment analysis
df['sentiment_score'] = df['Have you tried to get rid of your social media addiction? '].apply(lambda text: TextBlob(text).sentiment.polarity)

# Group data by age and gender and calculate the average sentiment score
sentiment_by_group = df.groupby(['What is your age?', 'What is your Gender'])['sentiment_score'].mean().reset_index()

# Display the results
display(sentiment_by_group)
```

## Data visualization

### Subtask:
Visualize the sentiment analysis results grouped by age and gender.


**Reasoning**:
Create visualizations of average sentiment by age and gender using the processed data.



```python
import matplotlib.pyplot as plt

# Visualization 1: Average Sentiment by Age
plt.figure(figsize=(12, 6))
plt.bar(sentiment_by_group['What is your age?'], sentiment_by_group['sentiment_score'], color='skyblue')
plt.xlabel('Age')
plt.ylabel('Average Sentiment Score')
plt.title('Average Sentiment Score by Age')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('sentiment_by_age.png')
plt.show()


# Visualization 2: Average Sentiment by Gender
plt.figure(figsize=(8, 6))
sentiment_by_gender = sentiment_by_group.groupby('What is your Gender')['sentiment_score'].mean().reset_index()
plt.bar(sentiment_by_gender['What is your Gender'], sentiment_by_gender['sentiment_score'], color=['lightcoral', 'lightgreen', 'lightskyblue'])
plt.xlabel('Gender')
plt.ylabel('Average Sentiment Score')
plt.title('Average Sentiment Score by Gender')
plt.tight_layout()
plt.savefig('sentiment_by_gender.png')
plt.show()
```
